```python
# ============================================================
#       TITANIC DATASET - EXPLORATORY DATA ANALYSIS
#       Register Number: 24EU02039
#       Google Colab - Single Code
# ============================================================

# Import Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Settings
pd.set_option('display.max_columns', None)
sns.set_style("whitegrid")

print("=" * 70)
print("       TITANIC DATASET - EXPLORATORY DATA ANALYSIS")
print("                 Register No: 24EU02039")
print("=" * 70)


# ============================================================
# 1. LOAD DATASET
# ============================================================

df = sns.load_dataset('titanic')

print("\n1. FIRST 5 ROWS")
print("-" * 50)
print(df.head())


# ============================================================
# 2. DATASET SHAPE
# ============================================================

print("\n2. DATASET SHAPE")
print("-" * 50)
print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])


# ============================================================
# 3. COLUMN NAMES
# ============================================================

print("\n3. COLUMN NAMES")
print("-" * 50)
print(df.columns.tolist())


# ============================================================
# 4. DATASET INFORMATION
# ============================================================

print("\n4. DATASET INFORMATION")
print("-" * 50)
df.info()


# ============================================================
# 5. STATISTICAL SUMMARY
# ============================================================

print("\n5. STATISTICAL SUMMARY")
print("-" * 50)
print(df.describe())


# ============================================================
# 6. CATEGORICAL DATA SUMMARY
# ============================================================

print("\n6. CATEGORICAL DATA SUMMARY")
print("-" * 50)
print(df.describe(include='object'))


# ============================================================
# 7. MISSING VALUES
# ============================================================

print("\n7. MISSING VALUES")
print("-" * 50)

missing = df.isnull().sum()
missing_percentage = (missing / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Values': missing,
    'Percentage (%)': missing_percentage.round(2)
})

print(missing_table.sort_values(
    by='Missing Values',
    ascending=False
))


# ============================================================
# 8. DUPLICATE ROWS
# ============================================================

print("\n8. DUPLICATE ROWS")
print("-" * 50)
print("Number of duplicate rows:", df.duplicated().sum())


# ============================================================
# 9. SURVIVAL ANALYSIS
# ============================================================

print("\n9. SURVIVAL ANALYSIS")
print("-" * 50)

print("\nSurvival Counts:")
print(df['survived'].value_counts())

print("\nSurvival Percentage:")
print(
    (df['survived'].value_counts(normalize=True) * 100).round(2)
)


# Graph 1 - Survival Count
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='survived')
plt.title('24EU02039 - Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()


# ============================================================
# 10. GENDER ANALYSIS
# ============================================================

print("\n10. GENDER ANALYSIS")
print("-" * 50)

print("Passengers by Gender:")
print(df['sex'].value_counts())

print("\nSurvival Rate by Gender:")
print(
    (df.groupby('sex')['survived'].mean() * 100).round(2)
)


# Graph 2 - Survival by Gender
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='sex', hue='survived')
plt.title('24EU02039 - Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()


# ============================================================
# 11. PASSENGER CLASS ANALYSIS
# ============================================================

print("\n11. PASSENGER CLASS ANALYSIS")
print("-" * 50)

print("Passengers in each class:")
print(df['pclass'].value_counts().sort_index())

print("\nSurvival Rate by Passenger Class:")
print(
    (df.groupby('pclass')['survived'].mean() * 100).round(2)
)


# Graph 3 - Survival by Passenger Class
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='pclass', hue='survived')
plt.title('24EU02039 - Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.show()


# ============================================================
# 12. AGE ANALYSIS
# ============================================================

print("\n12. AGE ANALYSIS")
print("-" * 50)

print("Average Age:", round(df['age'].mean(), 2))
print("Minimum Age:", df['age'].min())
print("Maximum Age:", df['age'].max())


# Graph 4 - Age Distribution
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', bins=30, kde=True)
plt.title('24EU02039 - Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()


# Graph 5 - Age vs Survival
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='survived', y='age')
plt.title('24EU02039 - Age Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.show()


# ============================================================
# 13. FARE ANALYSIS
# ============================================================

print("\n13. FARE ANALYSIS")
print("-" * 50)

print("Average Fare:", round(df['fare'].mean(), 2))
print("Minimum Fare:", df['fare'].min())
print("Maximum Fare:", df['fare'].max())


# Graph 6 - Fare Distribution
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='fare', bins=30, kde=True)
plt.title('24EU02039 - Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()


# Graph 7 - Fare by Passenger Class
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='pclass', y='fare')
plt.title('24EU02039 - Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()


# ============================================================
# 14. EMBARKATION ANALYSIS
# ============================================================

print("\n14. EMBARKATION ANALYSIS")
print("-" * 50)

print("Passengers by Embarkation Port:")
print(df['embarked'].value_counts())

print("\nSurvival Rate by Embarkation Port:")
print(
    (df.groupby('embarked')['survived'].mean() * 100).round(2)
)


# Graph 8 - Survival by Embarkation
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='embarked', hue='survived')
plt.title('24EU02039 - Survival by Embarkation Port')
plt.xlabel('Embarkation Port')
plt.ylabel('Number of Passengers')
plt.show()


# ============================================================
# 15. FAMILY SIZE ANALYSIS
# ============================================================

df['family_size'] = df['sibsp'] + df['parch'] + 1

print("\n15. FAMILY SIZE ANALYSIS")
print("-" * 50)

print("Family Size Counts:")
print(df['family_size'].value_counts().sort_index())

print("\nSurvival Rate by Family Size:")
print(
    (df.groupby('family_size')['survived'].mean() * 100).round(2)
)


# Graph 9 - Family Size Distribution
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='family_size')
plt.title('24EU02039 - Family Size Distribution')
plt.xlabel('Family Size')
plt.ylabel('Number of Passengers')
plt.show()


# Graph 10 - Survival by Family Size
plt.figure(figsize=(12, 6))
sns.barplot(
    data=df,
    x='family_size',
    y='survived'
)
plt.title('24EU02039 - Survival Rate by Family Size')
plt.xlabel('Family Size')
plt.ylabel('Survival Rate')
plt.show()


# ============================================================
# 16. GENDER + CLASS + SURVIVAL
# ============================================================

print("\n16. SURVIVAL BY GENDER AND CLASS")
print("-" * 50)

print(
    (df.groupby(['pclass', 'sex'])['survived'].mean() * 100)
    .round(2)
)


# Graph 11 - Gender and Class
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df,
    x='pclass',
    y='survived',
    hue='sex'
)
plt.title('24EU02039 - Survival Rate by Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()


# ============================================================
# 17. CORRELATION ANALYSIS
# ============================================================

print("\n17. CORRELATION MATRIX")
print("-" * 50)

numeric_df = df.select_dtypes(include=np.number)
correlation = numeric_df.corr()

print(correlation.round(2))


# Graph 12 - Correlation Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
plt.title('24EU02039 - Correlation Heatmap')
plt.show()


# ============================================================
# 18. OUTLIER DETECTION
# ============================================================

print("\n18. OUTLIER DETECTION")
print("-" * 50)

# Age Outliers
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['age'])
plt.title('24EU02039 - Age Outliers')

# Fare Outliers
plt.subplot(1, 2, 2)
sns.boxplot(y=df['fare'])
plt.title('24EU02039 - Fare Outliers')

plt.tight_layout()
plt.show()


# ============================================================
# 19. PAIRPLOT
# ============================================================

print("\n19. PAIRPLOT")

sns.pairplot(
    df[['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare']],
    hue='survived'
)

# Pairplot title cannot be set using plt.title() normally,
# so use suptitle.
plt.suptitle(
    '24EU02039 - Pairplot of Titanic Numerical Variables',
    y=1.02
)
plt.show()


# ============================================================
# 20. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("                       FINAL SUMMARY")
print("=" * 70)

print("\nDataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nOverall Survival Rate:")
print(
    round(df['survived'].mean() * 100, 2),
    "%"
)

print("\nSurvival Rate by Gender:")
print(
    (df.groupby('sex')['survived'].mean() * 100).round(2)
)

print("\nSurvival Rate by Passenger Class:")
print(
    (df.groupby('pclass')['survived'].mean() * 100).round(2)
)

print("\nAverage Age:")
print(round(df['age'].mean(), 2))

print("\nAverage Fare:")
print(round(df['fare'].mean(), 2))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# 21. CONCLUSIONS
# ============================================================

print("\n")
print("=" * 70)
print("                         CONCLUSIONS")
print("=" * 70)

print("""
1. The Titanic dataset contains passenger information such as
   age, gender, passenger class, fare, family members and survival.

2. The dataset contains missing values, especially in the age,
   deck and embarked columns.

3. Female passengers had a considerably higher survival rate
   than male passengers.

4. Passengers from higher classes generally had higher survival
   rates than passengers from lower classes.

5. Age is distributed across a wide range, with many passengers
   being young and middle-aged.

6. First-class passengers generally paid higher fares than
   second- and third-class passengers.

7. Fare and passenger class have a strong relationship.

8. Family size is associated with different survival patterns.

9. The correlation heatmap shows relationships between numerical
   variables.

10. Overall, gender and passenger class are important factors
    associated with survival in the Titanic dataset.
""")

print("=" * 70)
print("       TITANIC EDA COMPLETED SUCCESSFULLY")
print("                 24EU02039")
print("=" * 70)
```
